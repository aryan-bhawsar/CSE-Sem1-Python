```python
mylist = [10, 20, 30, 40]

print("Original list:", mylist)
print("1. Lenght:", len(mylist))
print("2. Add element:", mylist + [50])
print("3. Count 20:", mylist.count(20))
print("4. Index of 30:", mylist.index(30))

mylist.append(60)
print("5. Append 60:", mylist)

mylist.insert(1, 15)
print("6. Insert 15:", mylist)

mylist.remove(20)
print("7. Remove 20:", mylist)

mylist.sort()
print("8. Sort:", mylist)

mylist.reverse()
print("9. Reverse:", mylist)

print("10. Maximum:", max(mylist))
print("11. Minimum:", min(mylist))
```
